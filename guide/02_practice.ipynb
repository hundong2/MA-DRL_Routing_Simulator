# 응용: 큐 보상과 DDQN target

원본 산술 보상만 제한적으로 읽는다. DDQN target은 교육용 독립 구현이며 실제 학습 성공을 증명하지 않는다.

[학습 가이드](README.md)

```python
from pathlib import Path
import runpy
cwd = Path.cwd()
candidates = [cwd, cwd / 'guide', cwd / 'MA-DRL_Routing_Simulator' / 'guide']
guide = next(p for p in candidates if (p / 'examples' / 'reward_lab.py').exists())
lab = runpy.run_path(str(guide / 'examples' / 'reward_lab.py'))
reward = lab['load_queue_reward']()
print([(q, reward(q, 20)) for q in [0, 0.009, 0.02]])
assert abs(lab['ddqn_target'](1, 0.9, [1, 5], [100, 2]) - 2.8) < 1e-12
```

과제: online과 target의 최고 행동이 같을 때와 다를 때를 비교한다. 9 ms를 9초로 넣는 오류를 설명한다.
